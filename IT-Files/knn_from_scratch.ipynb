```python
from collections import Counter
import pandas as pd 
import numpy as np
import seaborn as sns 
```

```python
iris = sns.load_dataset("iris")

X = iris.drop(columns=["species"], axis=1).values 
y = iris["species"].map({"setosa":0, "versicolor":1, "virginica":2}).values
```

```python
def euclidian(x, y): 
    return np.sqrt(np.sum((x-y)**2))
```

```python
class Knn: 
    def __init__(self, k=3): 
        """
            args: 
                k : le nombre de voisin par defaut egale a 3 
        """
        self.k = k 
        
    def fit(self, X, y): 
        self.X = X 
        self.y = y 
    
    def __predict(self, new_obs): 
        distance = []
        for obs in self.X: 
            distance.append(euclidian(obs, new_obs))
        distance = np.array(distance)
        idx = np.argsort(distance)
        labels = y[idx][:self.k]
        return Counter(labels).most_common()[0][0]
    
    def predict(self, X): 
        return np.array([self.__predict(obs) for obs in X])
            
```

```python
new_obs = np.array([12.2, 3.4, 3, 1.7])
```

```python
model = Knn(k=6)
model.fit(X, y)
```

```python
pred = model.predict(X)
```

```python
(pred == y).sum() / len(y)
```
